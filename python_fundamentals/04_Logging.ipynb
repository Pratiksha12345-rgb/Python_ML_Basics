# 1. What is Logging?

Logging is a technique used in Python to record information about what is happening inside a program while it is running.

It helps developers track program execution, identify errors, understand warnings, and monitor application activities.

The Python logging module provides functionality to generate and manage log messages.

Why do we use Logging?
- To track program execution.
- To identify and debug errors.
- To monitor application activities.
- To record warning messages.
- To understand what happens when a program runs.
- To maintain records of application events.

Syntax

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("This is an information message")
```

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("User logged in")
logging.info("Application completed")
```

Code Explanation

- import logging – Imports Python's logging module.
- basicConfig() – Configures the logging system.
- level=logging.INFO – Sets the minimum logging level to INFO.
- logging.info() – Records an informational message.
- Each message describes an event in the application.

# 2. Why Use Logging?

# 3. print() vs logging

Both print() and logging are used to display information in Python.

However, they have different purposes.
- print() displays messages directly on the screen.
- logging records messages with different severity levels and can send them to the console or files.

Example

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

print("Application started")

logging.info("Application started")

logging.warning("Low disk space")

logging.error("Unable to connect to database")
```

Code Explanation
- print() – Displays a simple message.
- logging.info() – Records an informational message.
- logging.warning() – Records a warning.
- logging.error() – Records an error.
- format – Controls how log messages appear

# 5. Log Formatters

A Log Formatter is used to control the appearance and structure of log messages.

It allows us to include useful information such as:
- Date and time.
- Log level.
- Logger name.
- Log message.

The logging.Formatter() class is used to define the format of log messages.

Why do we use Log Formatters?
- To make log messages readable.
- To include timestamps.
- To identify the severity of messages.
- To organize log information.
- To make debugging easier.

Syntax

```python
formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)
```

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Low memory available")
logging.error("File loading failed")
```

Code Explanation
- logging.basicConfig() – Configures logging.
- level=logging.INFO – Displays INFO and higher-severity messages.
- %(asctime)s – Displays the date and time.
- %(levelname)s – Displays the log level.
- %(message)s – Displays the actual log message.
- logging.info() – Records an informational message.
- logging.warning() – Records a warning.
- logging.error() – Records an error.

Conclusion

Log formatters help organize log messages by including timestamps, log levels, and other useful information.

# 6. File Handlers

A File Handler is used to save log messages into a file instead of displaying them only on the screen.

Python provides the logging.FileHandler() class to write log messages into a file.

This is useful because developers can review the log records later, even after the program has finished running.

Why do we use File Handlers?
- To save log messages permanently.
- To maintain application history.
- To review errors later.
- To monitor application activities.
- To support debugging.

Syntax

```python
file_handler = logging.FileHandler("application.log")
```

```python
import logging

logger = logging.getLogger("MyApplication")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)

logger.addHandler(file_handler)

logger.info("Application started")
logger.warning("Low storage space")
logger.error("Unable to load file")
```

Code Explanation
- logging.getLogger() – Creates or retrieves a logger.
- logger.setLevel() – Sets the minimum logging level.
- logging.FileHandler() – Creates a handler to write logs into a file.
- logging.Formatter() – Defines the message format.
- setFormatter() – Applies the formatter to the file handler.
- addHandler() – Connects the handler to the logger.
- logger.info() – Writes an informational message.
- logger.warning() – Writes a warning message.
- logger.error() – Writes an error message.

Conclusion

File handlers allow Python applications to save log messages in files for future reference and debugging.

# 7. Console Handlers

A Console Handler is used to display log messages directly in the console or terminal.

Python provides the logging.StreamHandler() class to send log messages to a stream, usually the console.

Why do we use Console Handlers?
- To display log messages on the screen.
- To monitor application activities in real time.

Syntax

```python
console_handler = logging.StreamHandler()
```

```python
import logging

logger = logging.getLogger("ConsoleLogger")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.DEBUG)

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.debug("Debugging application")
logger.info("Application started")
logger.warning("Low disk space")
logger.error("An error occurred")
```

Code Explanation
- logging.getLogger() – Creates a logger.

- logging.StreamHandler() – Creates a console handler.
- console_handler.setLevel() – Sets the minimum level for the console.
- logging.Formatter() – Defines the log message format.
- setFormatter() – Applies the formatter.
- addHandler() – Connects the console handler to the logger.
- logger.debug() and other methods – Display messages in the console.

Conclusion

Console handlers help developers monitor application activities and view log messages directly on the screen.

# 8. Rotating Logs

Rotating Logs is a technique used to manage log files when they become large.

When an application runs continuously, it may generate many log messages. Over time, the log file can consume a large amount of storage.

Rotating logs automatically creates backup log files when the current log file reaches a specified size.

Python provides RotatingFileHandler from the logging.handlers module.

Why do we use Rotating Logs?
- To prevent log files from becoming too large.
- To manage storage space.
- To maintain backup log files.
- To organize application records.
- To support long-running applications.

Syntax

```python
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1024,
    backupCount=3
)
```

```python
import logging
from logging.handlers import RotatingFileHandler

# Create a fresh logger
logger = logging.getLogger("FreshRotatingLogger")
logger.setLevel(logging.INFO)
logger.propagate = False

# Remove old handlers
for h in logger.handlers[:]:
    h.close()
    logger.removeHandler(h)

# Create a new log file
handler = RotatingFileHandler(
    "rotation_demo.log",
    maxBytes=300,
    backupCount=3,
    delay=True
)

# Set log format
formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

# Generate log messages
for i in range(100):
    logger.info("This is log message number %d", i)

# Close the handler
logger.removeHandler(handler)
handler.close()

print("Rotating log example completed successfully!")
```

code Explanation

- RotatingFileHandler – Imports the rotating file handler.
- logging.getLogger() – Creates a logger.
- logger.setLevel(logging.INFO) – Sets the logging level.
- "application.log" – Specifies the main log file.
- maxBytes=1024 – Sets the maximum log file size to 1,024 bytes.
- backupCount=3 – Keeps up to three backup log files.
- handler.setFormatter() – Sets the log format.
- logger.addHandler() – Connects the rotating handler.
- for loop – Generates multiple log messages.
- When the file reaches its size limit, it rotates and creates backup files.

Conclusion

Rotating logs help manage large log files and maintain backup records without allowing one log file to grow indefinitely.

# 10. Difference Between File Handler and Console Handler

# File Handler	                 

Saves log messages in a file.

Uses FileHandler().

Useful for maintaining records.	  

Logs can be reviewed later.	      

# Console Handler

Displays log messages on the screen.

Uses StreamHandler().

Useful for real-time monitoring.

Logs are displayed during execution.


Example – Using Both Handlers Together

```python
import logging

logger = logging.getLogger("MyApp")
logger.setLevel(logging.INFO)

# Console Handler
console_handler = logging.StreamHandler()

# File Handler
file_handler = logging.FileHandler("myapp.log")

# Formatter
formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

# Add handlers
logger.addHandler(console_handler)
logger.addHandler(file_handler)

# Log messages
logger.info("Application started")
logger.warning("Low memory")
logger.error("Database connection failed")
```

The same messages are also saved in myapp.log.

# 11. Practical Exercise – Student Application Logging

Understand the Concept

This practical program demonstrates how logging can be used in a student registration application.

It records the registration process, student details, eligibility, and errors.

Example

```python
import logging

logging.basicConfig(
    filename="student.log",
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def register_student(name, age):

    logging.info("Student registration started")

    try:
        if age < 18:
            logging.warning("Student is below 18 years")

        else:
            logging.info("Student %s is registered", name)
            print("Registration successful!")

    except Exception:
        logging.exception("Registration failed")

    finally:
        logging.info("Registration process completed")

register_student("Vijay", 22)
register_student("Rahul", 16)

print("Program completed")
```

Code Explanation

- import logging – Imports the logging module.
- basicConfig() – Configures logging.
- filename="student.log" – Stores messages in a log file.
- level=logging.DEBUG – Enables all five standard logging levels.
- register_student() – Defines a student registration function.
- logging.info() – Records normal activities.
- logging.warning() – Records a warning for students below 18.
- logging.exception() – Records unexpected exception details.
- finally – Records that the registration process has finished.
- register_student() – Calls the function for two students.

Conclusion

This practical example demonstrates how logging can be used in real-world Python applications to monitor activities and record important events.
