# 1. Introduction to SQL

What is SQL?

SQL stands for Structured Query Language. It is used to create, manage, retrieve, and manipulate data stored in relational databases.

Why do we use SQL?
- To store and manage data.
- To retrieve specific information.
- To update and delete records.
- To connect related tables.
- To analyse data using queries.

# 2. Database Design

2.1 Database

A database is an organized collection of data.

```python
CREATE DATABASE sprint7_db;

USE sprint7_db;
```

2.2 Tables

A table stores data in rows and columns.

```python
CREATE TABLE departments (
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(50)
);
```

2.3 Primary Key

A primary key uniquely identifies each record in a table. It cannot contain duplicate or NULL values.

Example: dept_id.

2.4 Foreign Key

A foreign key connects one table to another table. It refers to the primary key of another table.

2.5 Relationships

Relationships connect tables using keys.

Types:
- One-to-One: One record is connected to one record.
- One-to-Many: One record is connected to many records.
- Many-to-Many: Many records are connected to many records through a junction table.

2.6 Constraints

Constraints are rules applied to table columns to maintain data accuracy.

Common constraints:
- PRIMARY KEY
- FOREIGN KEY
- NOT NULL
- UNIQUE
- CHECK
- DEFAULT

Example:

```python
CREATE TABLE employees (
    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE,
    salary DECIMAL(10,2) CHECK (salary > 0),
    dept_id INT,
    city VARCHAR(50) DEFAULT 'Bengaluru',
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
);
```

# 4. SQL Queries

4.1 SELECT

SELECT is used to retrieve data from a table.

```python
SELECT * FROM employees;
```

To select particular columns:

```python
SELECT emp_name, salary
FROM employees;
```

4.2 WHERE

WHERE filters records based on a condition.

```python
SELECT *
FROM employees
WHERE salary > 40000;
```

4.3 ORDER BY

ORDER BY sorts records in ascending or descending order.

```python
SELECT *
FROM employees
ORDER BY salary ASC;
```

Descending order:

```python
SELECT *
FROM employees
ORDER BY salary DESC;
```

4.4 GROUP BY

GROUP BY groups records that have the same values.

```python
SELECT dept_id, COUNT(*) AS total_employees
FROM employees
GROUP BY dept_id;
```

4.5 HAVING

HAVING filters grouped records.

```python
SELECT dept_id, AVG(salary) AS average_salary
FROM employees
GROUP BY dept_id
HAVING AVG(salary) > 40000;
```

4.6 DISTINCT

DISTINCT removes duplicate values from the result.

```python
SELECT DISTINCT city
FROM employees;
```

4.7 LIMIT

LIMIT restricts the number of records returned.

```python
SELECT *
FROM employees
LIMIT 3;
```

# 5. SQL Joins

A JOIN combines records from two or more tables using a related column.

5.1 INNER JOIN

Returns only matching records from both tables.

```python
SELECT e.emp_name, d.dept_name
FROM employees e
INNER JOIN departments d
ON e.dept_id = d.dept_id;
```

5.2 LEFT JOIN

Returns all records from the left table and matching records from the right table.

```python
SELECT e.emp_name, d.dept_name
FROM employees e
LEFT JOIN departments d
ON e.dept_id = d.dept_id;
```

5.3 RIGHT JOIN

Returns all records from the right table and matching records from the left table.

```python
SELECT e.emp_name, d.dept_name
FROM employees e
RIGHT JOIN departments d
ON e.dept_id = d.dept_id;
```

5.4 FULL JOIN

FULL OUTER JOIN returns all matching and non-matching records from both tables.

MySQL does not directly support FULL OUTER JOIN. We can use LEFT JOIN and RIGHT JOIN with UNION.

```python
SELECT e.emp_name, d.dept_name
FROM employees e
LEFT JOIN departments d
ON e.dept_id = d.dept_id

UNION

SELECT e.emp_name, d.dept_name
FROM employees e
RIGHT JOIN departments d
ON e.dept_id = d.dept_id;
```

5.5 CROSS JOIN

CROSS JOIN returns every possible combination of rows from both tables.

```python
SELECT e.emp_name, d.dept_name
FROM employees e
CROSS JOIN departments d;
```

# 6. Advanced SQL

6.1 UNION

UNION combines the results of two SELECT queries and removes duplicate records.

```python
SELECT city FROM employees
UNION
SELECT dept_name FROM departments;
```

6.2 CASE

CASE is used to apply conditional logic in SQL.

```python
SELECT emp_name, salary,
CASE
    WHEN salary >= 50000 THEN 'High Salary'
    WHEN salary >= 40000 THEN 'Medium Salary'
    ELSE 'Low Salary'
END AS salary_category
FROM employees;
```

6.3 Subqueries

A subquery is a query written inside another SQL query.

```python
SELECT emp_name, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
);
```

This query returns employees earning more than the average salary.

6.4 Common Table Expressions (CTEs)

A CTE is a temporary named result set that makes complex queries easier to read.

```python
WITH high_salary AS (
    SELECT *
    FROM employees
    WHERE salary > 40000
)
SELECT *
FROM high_salary;
```

6.5 Views

A view is a virtual table created using a SQL query.

```python
CREATE VIEW employee_details AS
SELECT emp_name, salary, city
FROM employees;
```

To access the view:

```python
SELECT * FROM employee_details;
```

6.6 Indexes

An index helps the database find records faster.

```python
CREATE INDEX idx_emp_name
ON employees(emp_name);
```


To remove an index:

```python
DROP INDEX idx_emp_name
ON employees;
```

6.7 Window Functions

Window functions perform calculations across related rows without combining them into one row.

A. ROW_NUMBER()

Assigns a unique sequential number to each row.

```python
SELECT emp_name, salary,
ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM employees;
```

B. RANK()

Assigns a rank to each row. Equal values receive the same rank, and the next rank is skipped.

```python
SELECT emp_name, salary,
RANK() OVER (ORDER BY salary DESC) AS employee_rank
FROM employees;
```

C. DENSE_RANK()

Assigns ranks without skipping the next rank after a tie.

```python
SELECT emp_name, salary,
DENSE_RANK() OVER (ORDER BY salary DESC) AS employee_rank
FROM employees;
```

D. LAG()

Returns a value from the previous row.

```python
SELECT emp_name, salary,
LAG(salary) OVER (ORDER BY emp_id) AS previous_salary
FROM employees;
```

E. LEAD()

Returns a value from the next row.

```python
SELECT emp_name, salary,
LEAD(salary) OVER (ORDER BY emp_id) AS next_salary
FROM employees;
```
