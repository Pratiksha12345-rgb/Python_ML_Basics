# 1. Introduction to Object-Oriented Programming

What is OOP?

Object-Oriented Programming (OOP) is a programming approach that organizes code using classes and objects.

It helps us write reusable, organized, and maintainable code.

OOP is based on four main principles:
    1. Encapsulation
    2. Abstraction
    3. Inheritance
    4. Polymorphism
    
Real-world example:

A car has properties such as color, brand, and speed. It also has behaviors such as starting, stopping, and accelerating.

In Python, we can represent a car using a class and create different car objects.

# 2. Class Definition:

A class is a blueprint or template used to create objects. It defines the properties and methods that its objects can have.

Example: A Student class can define student name, age, and marks.

```python
#Implementation:

class Student:
    name = "Pratiksha"
    age = 23
    course = "MCA"

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
```

# 3. Object Definition:

An object is an instance of a class. It contains data and can use the methods defined inside the class.

Example: If Student is a class, Pratiksha and Rahul can be two different student objects.

```python
# Implementation:

class Student:
    def display(self):
        print("Welcome to Python OOP")

student1 = Student()
student2 = Student()

student1.display()
student2.display()
```

# Explanation:
- Student() creates an object.
- student1 and student2 are two separate objects.
- Both objects can access the class method.

# 4. Constructor (__init__) Definition:

A constructor is a special method that runs automatically when an object is created. In Python, __init__() initializes the object's attributes.

Example: When a student is registered, their name and age are initialized.

```python
# Implementation:

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

s1 = Student("Pratiksha", 23)
s1.display()
```

Explanation:
- __init__() executes automatically.
- self.name stores the student's name.
- self.age stores the student's age.
- Constructor helps initialize object data.

# 5. Instance Variables Definition:

Instance variables are variables that belong to individual objects. Every object can have different values.

```python
#Implementation:

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

e1 = Employee("Pratiksha", 25000)
e2 = Employee("Rahul", 30000)

print(e1.name, e1.salary)
print(e2.name, e2.salary)
```

Explanation: Each employee object stores its own name and salary.

# 6. Class Variables Definition:

Class variables are shared by all objects of a class. They are defined inside the class but outside instance methods.

```python
# Implementation:

class Employee:
    company = "Google"

    def __init__(self, name):
        self.name = name

e1 = Employee("Pratiksha")
e2 = Employee("Rahul")

print(e1.company)
print(e2.company)
```

Explanation: Both employees share the same company name.

# 7. Instance Methods

Definition:

Instance methods are functions defined inside a class that work with individual object data. They use self as the first parameter.

```python
#Implementation:

class Calculator:
    def add(self, a, b):
        return a + b

c1 = Calculator()

result = c1.add(10, 20)
print("Addition:", result)
```

# 8. Static Methods

Definition:

A static method is a method that does not require instance or class-specific data. It is created using the @staticmethod decorator.

```python
#Implementation:

class Calculator:

    @staticmethod
    def multiply(a, b):
        return a * b

print(Calculator.multiply(5, 4))
```

Explanation: Static methods can be called directly using the class name.

# 9. Class Methods

Definition:

A class method works with class-level data. It uses cls as its first parameter and is created using @classmethod.

```python
#Implementation:

class Employee:
    company = "Google"

    @classmethod
    def change_company(cls, name):
        cls.company = name

Employee.change_company("Google")

print(Employee.company)
```

Explanation: The class method changes the shared company name.

# 10. Properties (@property)

Definition:

The @property decorator allows us to access a method like an attribute. It is useful for controlling access to object data.

```python
#Implementation:

class Student:
    def __init__(self, marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):
        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid marks")

s1 = Student(85)

print(s1.marks)

s1.marks = 90
print(s1.marks)
```

Explanation:

- @property reads the marks.
- @marks.setter controls how marks are updated.
- It prevents invalid values.

# 11. Encapsulation

Definition:

Encapsulation means combining data and methods inside a class and controlling access to the data.

Python uses naming conventions and name mangling to support data protection.

```python
#Implementation:

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

account.deposit(2000)

print("Balance:", account.get_balance())
```

Explanation:

- __balance is a name-mangled attribute.
- deposit() updates the balance.
- get_balance() provides controlled access.

Real-world example: A bank customer can deposit money without directly changing the bank's internal account balance.

# 12. Abstraction

Definition:

Abstraction means hiding internal implementation details and showing only the necessary functionality.

Python supports abstraction using the abc module.

```python
#Implementation:

from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car started")

car = Car()
car.start()
```

Explanation:

- ABC creates an abstract base class.
- @abstractmethod defines a method that subclasses must implement.
- The Car class provides the actual implementation.

# 13. Inheritance

Definition:

Inheritance allows one class to acquire the properties and methods of another class.

The existing class is called the parent class, and the new class is called the child class.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()

dog.eat()
dog.bark()
```

Explanation: Dog inherits the eat() method from Animal.

# 14. Multiple Inheritance

Definition:

Multiple inheritance means one child class inherits from more than one parent class.

```python
class Father:
    def father_skill(self):
        print("Father knows driving")

class Mother:
    def mother_skill(self):
        print("Mother knows cooking")

class Child(Father, Mother):
    pass

c = Child()

c.father_skill()
c.mother_skill()
```

Explanation: Child inherits methods from both Father and Mother.

# 15. Multilevel Inheritance

Definition:

Multilevel inheritance occurs when a class inherits from another child class, creating multiple levels of inheritance.

```python
class Grandfather:
    def house(self):
        print("Grandfather owns a house")

class Father(Grandfather):
    def car(self):
        print("Father owns a car")

class Son(Father):
    def bike(self):
        print("Son owns a bike")

s = Son()

s.house()
s.car()
s.bike()
```

Explanation: Son inherits features from Father and Grandfather.

# 16. Hierarchical Inheritance

Definition:

Hierarchical inheritance occurs when multiple child classes inherit from the same parent class.

```python
class Animal:
    def eat(self):
        print("Animals eat food")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

d = Dog()
c = Cat()

d.eat()
d.bark()

c.eat()
c.meow()
```

Explanation: Both Dog and Cat inherit the eat() method from Animal.

# 17. Polymorphism

Definition:

Polymorphism means one interface or method name can perform different actions depending on the object.

```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

def animal_sound(animal):
    animal.sound()

d = Dog()
c = Cat()

animal_sound(d)
animal_sound(c)
```

Explanation: The same sound() method produces different results for different objects.

# 18. Method Overriding

Definition:

Method overriding occurs when a child class defines a method with the same name as a method in its parent class.

```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

d = Dog()
d.sound()
```

Explanation: The child class overrides the parent class method.

# 19. Method Overloading (Python Approach)

Definition:

Method overloading means using the same method name with different numbers or types of arguments.

Python does not support traditional method overloading based on method signatures. We can achieve similar behavior using default arguments or *args.

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c

c = Calculator()

print(c.add(10))
print(c.add(10, 20))
print(c.add(10, 20, 30))
```


Explanation: The same method works with one, two, or three arguments.

# 20. Composition

Definition:

Composition is a relationship in which one class contains and owns another object as an essential part of itself.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")

car = Car()
car.start_car()
```

Explanation: A Car creates and owns its Engine object.

21. Aggregation

Definition:

Aggregation is a relationship in which one class uses another class's object, but the contained object can exist independently.

```python
class Employee:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, employee):
        self.employee = employee

emp = Employee("Pratiksha")

dept = Department(emp)

print(dept.employee.name)
```

Explanation: Employee can exist independently even if the Department object is removed.

# 22. Association

Definition:

Association represents a relationship between two independent classes that interact with each other.

```python
class Student:
    def __init__(self, name):
        self.name = name

class Teacher:
    def teach(self, student):
        print("Teacher is teaching", student.name)

s = Student("Pratiksha")
t = Teacher()

t.teach(s)
```

Explanation: Teacher and Student are independent objects that interact with each other.

# 3. Magic (Dunder) Methods

Definition:

Magic methods are special methods in Python that have double underscores before and after their names.

They allow us to define how objects behave with built-in operations.

Examples:
- __init__()
- __str__()
- __repr__()
- __len__()
- __add__()

```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return self.title

book = Book("Python Programming")

print(book)
```

Output:
Python Programming

Explanation: __str__() controls the readable string representation of an object.

# 24. super()

Definition:

super() is used to access methods from a parent class inside a child class.

```python
class Parent:
    def display(self):
        print("This is parent class")

class Child(Parent):
    def display(self):
        super().display()
        print("This is child class")

c = Child()
c.display()
```


Explanation: super() calls the parent class implementation of display().

# 25. __str__()

Definition:

__str__() returns a human-readable string representation of an object. It is used when we print an object.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"

s = Student("Pratiksha")

print(s)
```

26. __repr__()
Definition:

__repr__() returns an official or developer-oriented representation of an object. It is useful for debugging.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student(name={self.name!r})"

s = Student("Pratiksha")

print(repr(s))
```

Explanation: repr() displays the developer-friendly representation of the object.

# 27. __len__()

Definition:

__len__() defines what should happen when the built-in len() function is called on an object.

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team = Team(["Pratiksha", "Rahul", "Sneha"])

print("Total Members:", len(team))
```

Explanation: __len__() returns the number of members in the team.

# 28. Dataclasses (Bonus)

Definition:

A dataclass is a special Python class used mainly to store data. It automatically generates methods such as __init__() and __repr__().

It is available through the dataclasses module.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    course: str

s = Student("Pratiksha", 23, "MCA")

print(s)
```

Explanation: The @dataclass decorator automatically creates common methods, reducing the amount of code we need to write.

# 29. Singleton Pattern (Bonus)

Definition:

The Singleton Pattern is a design pattern that ensures a class has only one shared instance.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance

obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)
```

Explanation:

- __new__() controls object creation.
- The first call creates an instance.
- Later calls return the same instance.
- is checks whether both variables refer to the same object.

Real-world example: A shared application configuration manager can use a Singleton pattern

# 34. OOP in AI/ML Engineering

OOP is useful in AI/ML because it helps organize machine learning code into reusable components.

```python
#example

class MLModel:
    def __init__(self, model_name):
        self.model_name = model_name

    def train(self):
        print(self.model_name, "model is training")

    def predict(self):
        print(self.model_name, "model is predicting")

model = MLModel("Random Forest")

model.train()
model.predict()
```

Explanation:

- MLModel represents a machine learning model.
- train() represents the training process.
- predict() represents the prediction process.
- Different model objects can be created using the same class structure.
