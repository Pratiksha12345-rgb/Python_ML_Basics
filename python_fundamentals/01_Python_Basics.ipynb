# 1. Variables and Data Types
Definition:
A variable is a name used to store a value in Python. A data type defines what kind of value is stored in a variable.

# Types of Data Types:
    1. Integer (int) – Whole numbers.
    2. Float (float) – Decimal numbers.
    3. String (str) – Text values.
    4. Boolean (bool) – True or False.
    5. List (list) – Ordered and changeable collection.
    6. Tuple (tuple) – Ordered and unchangeable collection.
    7. Set (set) – Collection of unique values.
    8. Dictionary (dict) – Stores data in key-value pairs.

```python
# Variables and Data Types

name = "Pratiksha"
age = 23
height = 5.4
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is Student:", is_student)

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

2. Type Casting
Definition:
Type casting means converting one data type into another data type.
Types:
    1. int() – Converts a value into an integer.
    2. float() – Converts a value into a float.
    3. str() – Converts a value into a string.
    4. bool() – Converts a value into a Boolean.

```python
# Type Casting

a = "10"
b = int(a)

print("String to Integer:", b)
print(type(b))

x = 20
y = float(x)

print("Integer to Float:", y)
print(type(y))

number = 100
text = str(number)

print("Integer to String:", text)
print(type(text))

value = 1
result = bool(value)

print("Integer to Boolean:", result)
print(type(result))
```

3. Python Operators
Definition:
Operators are special symbols used to perform operations on variables and values.

Types of Operators:

    1. Arithmetic Operators
    2. Comparison Operators
    3. Assignment Operators
    4. Logical Operators
    5. Bitwise Operators
    6. Membership Operators
    7. Identity 


A. Arithmetic Operators
Arithmetic operators are used to perform mathematical operations.

```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

B. Comparison Operators
Comparison operators compare two values and return True or False.

```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

C. Assignment Operators
Assignment operators are used to assign or update values in variables.

```python
a = 10

a += 5
print("Addition:", a)

a -= 3
print("Subtraction:", a)

a *= 2
print("Multiplication:", a)

a /= 4
print("Division:", a)
```

D. Logical Operators
Logical operators are used to combine multiple conditions.

```python
a = 10
b = 5

print(a > 5 and b < 10)
print(a > 15 or b < 10)
print(not(a > 5))
```

E. Bitwise Operators
Bitwise operators perform operations on the binary representation of numbers.

```python
a = 5
b = 3

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```

F. Membership Operators
Membership operators check whether a value exists in a collection.

```python
names = ["Pratiksha", "Ajay", "Ravi"]

print("Ajay" in names)
print("Ramesh" not in names)
```

G.Identity Operators
Identity operators check whether two variables refer to the same object in memory.

```python
a = [10, 20, 30]
b = a
c = [10, 20, 30]

print(a is b)
print(a is c)
print(a is not c)
```

8. Break, Continue and Pass
Definition:
These are control statements used to change the normal execution of loops.

    A. Break Statement
    The break statement stops the loop immediately.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

B. Continue Statement
The continue statement skips the current iteration and moves to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

C. Pass Statement
The pass statement does nothing. It is used as a placeholder when a statement is required syntactically.

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

9. Python Collections
Definition:
Collections are used to store multiple values in a single variable.
Types:
    1. List
    2. Tuple
    3. Set
    4. Dictionary

A. List
A list is an ordered and changeable collection. It allows duplicate values.

```python
students = ["Pratiksha", "Ajay", "Ravi", "Sneha"]

print(students)

students.append("Rahul")
print("After Append:", students)

students.remove("Ajay")
print("After Remove:", students)

print("First Student:", students[0])
print("Length:", len(students))
```

B. Tuple
A tuple is an ordered and unchangeable collection. It allows duplicate values.

```python
student = ("Pratiksha", 23, "MCA")

print(student)
print("Name:", student[0])
print("Age:", student[1])
print("Course:", student[2])
print("Length:", len(student))
```

C. Set
A set is an unordered collection of unique elements. It does not allow duplicate values.

```python
numbers = {10, 20, 30, 20, 40}

print("Set:", numbers)

numbers.add(50)
print("After Add:", numbers)

numbers.remove(20)
print("After Remove:", numbers)
```

D. Dictionary
A dictionary stores data in key-value pairs.

```python
student = {
    "name": "Pratiksha",
    "age": 23,
    "course": "MCA"
}

print(student)

print("Name:", student["name"])

student["age"] = 24
print("Updated Dictionary:", student)

student["city"] = "Belgaum"
print("After Adding:", student)
```

12. Functions in Python
Definition:
A function is a reusable block of code that performs a particular task. Functions help reduce code repetition.

```python
def greet():
    print("Welcome to Python Programming")

greet()
```

```python
#A. Function with Arguments

def add(a, b):
    print("Addition:", a + b)

add(10, 20)
```

```python
#B. Function with Return Value

def multiply(a, b):
    return a * b

result = multiply(5, 4)

print("Multiplication:", result)
```

```python
# C. Default Arguments: A default argument has a predefined value. It is used when the caller does not provide that argument.

def greet(name="Pratiksha"):
    print("Hello", name)

greet()
greet("Ajay")
```

```python
# D. Keyword Arguments: Keyword arguments pass values using parameter names.

def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=23, name="Pratiksha")
```

```python
#E. Variable-Length Arguments (*args):*args allows a function to accept multiple positional arguments.

def add_numbers(*args):
    total = 0

    for number in args:
        total += number

    print("Total:", total)

add_numbers(10, 20, 30, 40)
```

```python
#F. Keyword Variable-Length Arguments (**kwargs):**kwargs allows a function to accept multiple keyword arguments.

def student_details(**kwargs):
    print(kwargs)

student_details(name="Pratiksha", age=23, course="MCA")
```

```python
#G. Lambda Function: A lambda function is a small anonymous function written in a single expression.

square = lambda x: x * x

print("Square:", square(5))
```

```python
#H. Recursion: Recursion is a process in which a function calls itself.

def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print("Factorial:", factorial(5))
```

14. Generators in Python
Definition:
A generator is a special type of function that produces values one at a time using the yield keyword.

```python
def my_generator():
    yield 10
    yield 20
    yield 30

for value in my_generator():
    print(value)
```

15. Yield Keyword
Definition:
The yield keyword returns a value from a generator function while preserving its execution state.

```python
def count_numbers():
    yield 1
    yield 2
    yield 3

result = count_numbers()

print(next(result))
print(next(result))
print(next(result))
```

16. Decorators in Python
Definition:
A decorator is a function that adds extra functionality to another function without changing its original code.

```python
def my_decorator(func):

    def wrapper():
        print("Before Function")

        func()

        print("After Function")

    return wrapper


@my_decorator
def greet():
    print("Hello Python")


greet()
```

17. Closures in Python
Definition:
A closure is a function that remembers variables from its outer function even after the outer function has finished executing.

```python
def multiplier(x):

    def multiply(number):
        return number * x

    return multiply


double = multiplier(2)

print(double(5))
print(double(10))
```

18. Global and Nonlocal Variables
Definition:
Global and nonlocal keywords are used to access and modify variables from different scopes.

```python
#A. Global Keyword: The global keyword allows a function to modify a variable defined at the module level.

x = 10

def change_value():
    global x
    x = 20

change_value()

print(x)
```

```python
#B. Nonlocal Keyword: The nonlocal keyword allows an inner function to modify a variable in its nearest enclosing function.

def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()

    print("Value:", x)

outer()
```

23. Built-in Functions in Python
Definition:
Built-in functions are predefined functions available in Python. We can use them without importing any module.

A. map()
The map() function applies a given function to every element in an iterable.

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

print(result)
```

B. filter()
The filter() function selects elements that satisfy a given condition.

```python
numbers = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

C. zip()
The zip() function combines elements from two or more iterables.

```python
names = ["Pratiksha", "Ajay", "Ravi"]
ages = [23, 24, 25]

result = list(zip(names, ages))

print(result)
```

D. enumerate()
The enumerate() function returns both the index and value of each element.

```python
students = ["Pratiksha", "Ajay", "Ravi"]

for index, name in enumerate(students):
    print(index, name)
```

E. sorted()
The sorted() function returns a new sorted list.

```python
numbers = [5, 2, 8, 1, 3]

result = sorted(numbers)

print(result)
```

F. any()
The any() function returns True if at least one element in an iterable is True.

```python
numbers = [0, 0, 5, 0]

print(any(numbers))
```

G. all()
The all() function returns True only when all elements in an iterable are True.

```python
numbers = [1, 2, 3, 4]

print(all(numbers))

numbers = [1, 0, 3, 4]

print(all(numbers))
```

H. isinstance()
The isinstance() function checks whether an object belongs to a particular data type or class.

```python
age = 23
name = "Pratiksha"

print(isinstance(age, int))
print(isinstance(name, str))
print(isinstance(age, str))
```

I. id()
The id() function returns the identity of an object during its lifetime.

```python
a = [10, 20, 30]
b = a

print(id(a))
print(id(b))
print(id(a) == id(b))
```
