# 1. Introduction to Exception Handling

Understand the Concept

Exception Handling is a way to manage errors in a Python program so that the program does not stop unexpectedly.

An exception is an error that occurs during program execution.

Examples:

- Dividing a number by zero.
- Entering invalid input.
- Accessing a missing file.
- Using an undefined variable.

With exception handling, we can catch these errors and respond gracefully instead of allowing the program to crash.

Why do we use Exception Handling?

- To prevent sudden program termination.
- To handle runtime errors.
- To display user-friendly error messages.
- To make programs more reliable.
- To improve debugging.

Example

```python
num1 = 10
num2 = 0

result = num1 / num2

print("Result:", result)
```

# 2. Try Block

Understand the Concept

The try block is used for exception handling.

It allows us to test a block of code that may produce an error.

If an error occurs inside the try block, Python moves to the matching except block instead of immediately stopping the program.

Why do we use Try?

- To test risky code.
- To identify possible runtime errors.
- To handle errors using the except block.
- To prevent unexpected program termination when errors are handled.

Syntax

try:

    # Code that may cause an error

except:

    # Code to handle the error

Example

```python
try:
    num1 = 10
    num2 = 2

    result = num1 / num2

    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero!")
```

# 3. Except Block

Understand the Concept

The except block is part of Python's exception-handling system.

It works together with the try block.

If an error occurs inside the try block, Python moves to the matching except block to handle the error.

Each except block can handle a specific type of error, such as ZeroDivisionError, ValueError, or TypeError.

Why do we use Except?

- To handle runtime errors.
- To prevent the program from crashing.
- To display meaningful error messages.
- To handle different types of exceptions separately.

Syntax

try:
    # Risky code

except ErrorType:

    # Code to handle the error

Example

```python
try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))

    result = num1 / num2

    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid numbers!")
```

# 4. Else Block

Understand the Concept

The else block is used together with try and except.

It runs only when no exception occurs inside the try block.

Think of it as:

- try – Executes risky code.
- except – Handles errors.
- else – Executes when everything works successfully.

Why do we use Else?

- To execute code only when no error occurs.
- To separate successful operations from error-handling code.
- To make programs easier to understand.



```python
try:
    num1 = 10
    num2 = 2

    result = num1 / num2

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

else:
    print("Result:", result)
    print("Division successful! No errors occurred.")
```

Code Explanation

- try – Performs the division.
- except – Handles division errors.
- else – Runs only if the division is successful.
- print() – Displays the result and success message.

# 5. Finally Block

The finally block is used with exception handling.

It always executes whether an exception occurs or not.

It is commonly used for cleanup tasks, such as:

- Closing files.
- Releasing resources.
- Disconnecting databases.
- Cleaning up temporary resources.

Why do we use Finally?

- To execute important cleanup code.
- To close files safely.
- To release resources.
- To ensure cleanup code runs even when an error occurs.


```python
file = None

try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)

except FileNotFoundError:
    print("Error: File not found!")

finally:
    if file is not None:
        file.close()
        print("File closed successfully.")
    else:
        print("File was never opened.")

    print("Program finished.")
```

# 6. Raise Keyword

The raise keyword is used to manually generate an exception in Python.

Normally, exceptions occur automatically when an error happens.

But sometimes we want to force an error when a particular condition is not satisfied.

This is useful for validation and custom error handling.

Why do we use Raise?

- To generate exceptions manually.
- To validate user input.
- To enforce business rules.
- To display meaningful error messages.

Syntax

raise ExceptionType("Error message")

```python
def register_student(name, age):
    if age < 5:
        raise ValueError("Age must be at least 5 to register.")

    print(f"Student {name} registered successfully!")


try:
    register_student("Vijay", 3)

except ValueError as e:
    print("Registration failed:", e)
```

# 7. Custom Exceptions

Understand the Concept

Python provides built-in exceptions such as ValueError, TypeError, and ZeroDivisionError.

But sometimes we need to create our own exception types for special situations in a program.

A custom exception is a user-defined exception created by inheriting from Python's built-in Exception class.

Why do we use Custom Exceptions?
- To create specific error messages.
- To handle application-specific errors.
- To improve code readability.
- To make debugging easier.
- To manage errors in large projects.

Syntax

class MyCustomError(Exception):

    pass

```python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        super().__init__(
            f"Balance {balance} is not enough to withdraw {amount}"
        )


def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)

    balance -= amount

    print("Withdrawal successful!")
    print("Remaining balance:", balance)

    return balance


try:
    withdraw(1000, 1500)

except InsufficientFundsError as e:
    print("Transaction failed:", e)
```

# 19. Practical Program – Exception Handling with User Input

Understand the Concept

This program demonstrates how to use try, except, else, and finally together in a real-world example.

It accepts two numbers from the user and performs division while handling possible errors.

```python
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))

    result = num1 / num2

except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

except ValueError:
    print("Error: Please enter valid numbers!")

else:
    print("Result:", result)
    print("Division successful!")

finally:
    print("Program finished.")
```
