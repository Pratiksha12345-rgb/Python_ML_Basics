# 1. Introduction to File Handling

File handling in Python is used to create, read, write, update, and delete files.

It helps us store and manage data permanently.

Why do we use file handling?
- To store information in files.
- To read existing data.
- To update and delete files.
- To work with CSV and JSON data.
- To manage folders and directories.

# 2. Reading Text Files

Reading a text file means opening a file and retrieving its contents.

```python
from pathlib import Path

file_path = Path("sample.txt")

file_path.write_text("Welcome to Python File Handling!")

with open(file_path, "r") as file:
    content = file.read()

print(content)
```

Explanation:
- open() opens the file.
- "r" means read mode.
- read() reads the complete file.
- with automatically closes the file.

# 3. Writing Text Files

Writing means storing information inside a file.

```python
with open("sample.txt", "w") as file:
    file.write("Python is easy to learn.")

print("Data written successfully!")
```

Explanation:
- "w" means write mode.
- write() stores text in the file.
- Existing content is replaced in write mode.

# 4. CSV Files

CSV stands for Comma-Separated Values.

It is used to store tabular data in rows and columns.

```python
import csv

data = [
    ["Name", "Age", "Course"],
    ["Pratiksha", 23, "Python"],
    ["Rahul", 24, "SQL"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created successfully!")
```

Reading CSV file:

```python
with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

# 5. JSON Files

JSON stands for JavaScript Object Notation.

It is used to store and exchange structured data.

```python
import json

student = {
    "name": "Pratiksha",
    "age": 23,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created successfully!")
```

Reading JSON file:

```python
with open("student.json", "r") as file:
    data = json.load(file)

print(data)
```

Explanation:
- json.dump() writes data into a JSON file.
- json.load() reads data from a JSON file.
- indent=4 makes the JSON file easier to read.

# 6. Pickle Files

Pickle is a Python module used to save and load Python objects in binary format.

```python
import pickle

student = {
    "name": "Pratiksha",
    "marks": 85
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

print("Pickle file saved successfully!")
```

Reading Pickle file:

```python
with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

Explanation:
- pickle.dump() saves Python objects.
- pickle.load() loads saved objects.
- "wb" means write binary.
- "rb" means read binary.

Important: Only load pickle files from trusted sources because loading untrusted pickle files can execute harmful code.

# 7. Context Manager (with)

A context manager manages resources automatically.

The with statement automatically closes a file after the operation is completed.

```python
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)

print("File closed automatically.")
```

# 8. pathlib

pathlib is a Python module used to work with file and directory paths.

```python
from pathlib import Path

file_path = Path("sample.txt")

print("File name:", file_path.name)
print("File exists:", file_path.exists())
print("File suffix:", file_path.suffix)
```

# 9. os Module

The os module provides functions to interact with the operating system.

It helps manage files, folders, and directories.

```python
import os

print("Current directory:", os.getcwd())
print("Files in current directory:", os.listdir())
```

# 10. shutil Module

The shutil module is used to copy, move, and delete files and directories.

```python
import shutil

shutil.copy("sample.txt", "sample_copy.txt")

print("File copied successfully!")
```

Explanation:

shutil.copy() creates a copy of the source file at the destination.

# 11. File Validation

File validation means checking whether a file exists and whether it has the correct file extension.



```python
from pathlib import Path

file_path = Path("sample.txt")

if file_path.exists() and file_path.is_file():
    if file_path.suffix == ".txt":
        print("File validation successful!")
    else:
        print("Invalid file extension.")
else:
    print("File does not exist.")
```

# 12. Directory Operations

Directory operations are used to create, check, and manage folders.

```python
from pathlib import Path

folder = Path("My_Files")

folder.mkdir(exist_ok=True)

print("Directory created successfully!")
print("Directory exists:", folder.exists())
```

Creating a file inside the directory:

```python
file_path = folder / "example.txt"

file_path.write_text("This file is inside My_Files folder.")

print("File created inside directory!")
```

# 13. Practical Example – Student File Management

This program creates a student file, writes student details, and reads the information.

```python
from pathlib import Path

folder = Path("Student_Records")
folder.mkdir(exist_ok=True)

file_path = folder / "student.txt"

with open(file_path, "w") as file:
    file.write("Name: Pratiksha\n")
    file.write("Course: Python\n")
    file.write("Status: Learning")

print("Student details saved successfully!")

with open(file_path, "r") as file:
    print("\nStudent Details:")
    print(file.read())
```
