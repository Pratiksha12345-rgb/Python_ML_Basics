# 1. Introduction to NumPy Random Module

NumPy's random module is used to generate random numbers and random samples.

It helps us create random data for simulations, testing, experiments, Data Analysis and Machine Learning.

The NumPy random module provides functions for:
- Random Numbers
- Random Seed
- Random Integers
- Random Choice
- Shuffle
- Permutation
- Random Distributions

# 2. Import NumPy

```python
import numpy as np

print("NumPy imported successfully")
```

# 3. Random Numbers

Random numbers are numbers generated without a fixed predictable sequence, depending on the random generator and its state.

NumPy can generate random numbers between 0 and 1 using np.random.rand().

```python
random_numbers = np.random.rand(5)

print(random_numbers)
```

Interpretation
- np.random.rand(5) generates 5 random floating-point numbers.
- These numbers are generally between 0 and 1, excluding 1.

# 4. Random Seed

A random seed is a starting value used by a random number generator.

Setting a seed helps generate the same random sequence again, making experiments reproducible.

```python
np.random.seed(42)

numbers = np.random.rand(5)

print(numbers)
```

Run the same code again

```python
np.random.seed(42)

numbers = np.random.rand(5)

print(numbers)
```

Interpretation
- np.random.seed(42) sets the seed.
- Running the same random operation with the same seed reproduces the same values.
- This helps in testing, experiments and Machine Learning model development.

Best practice for new NumPy code: 
use np.random.default_rng(seed) instead of the legacy global np.random.seed() approach. Both are demonstrated in this notebook.

# 5. Random Integers – np.random.randint()
np.random.randint() generates random integer values within a specified range.

Syntax

np.random.randint(low, high, size)

- low – starting value (included).

- high – ending value (excluded).

- size – number of random integers.

```python
random_integers = np.random.randint(1, 101, size=5)

print(random_integers)
```

Practical Example

Generate random marks for five students.

```python
student_marks = np.random.randint(35, 101, size=5)

print("Student Marks:", student_marks)
```

# 6. Random Choice – np.random.choice()

np.random.choice() selects random elements from a given array.

It can be used for random selection, sampling and simulation.

```python
students = np.array(["A", "B", "C", "D", "E"])

selected_student = np.random.choice(students)

print("Selected Student:", selected_student)
```

Select Multiple Students

```python
students = np.array(["A", "B", "C", "D", "E"])

selected_students = np.random.choice(students, size=3)

print(selected_students)
```

Select Without Replacement

```python
students = np.array(["A", "B", "C", "D", "E"])

selected_students = np.random.choice(students, size=3, replace=False)

print(selected_students)
```

Interpretation
- size=3 selects three students.
- replace=False prevents selecting the same student more than once in the selection.

# 7. Shuffle – np.random.shuffle()

Shuffle randomly rearranges the elements of an array.

np.random.shuffle() modifies the original array.

```python
marks = np.array([85, 90, 78, 92, 88])

np.random.shuffle(marks)

print(marks)
```

Interpretation
The original array is rearranged randomly. The values remain the same, but their order changes.

Important
np.random.shuffle() changes the original array directly. It does not return a new shuffled array.

# 8. Permutation – np.random.permutation()

Permutation randomly rearranges the elements of an array but returns a new array without modifying the original array.

```python
marks = np.array([85, 90, 78, 92, 88])

shuffled_marks = np.random.permutation(marks)

print("Original Array:", marks)
print("Permuted Array:", shuffled_marks)
```

Interpretation
- marks remains unchanged.
- shuffled_marks contains the same elements in a random order.

# Difference Between Shuffle and Permutation

# Shuffle	Permutation
Modifies the original array	--> Returns a new array

Does not return a shuffled array as its main result -->	Returns a permuted array

Used to rearrange original data	--> Used to create a rearranged copy


# 9. Random Distributions

A random distribution describes how random values are generated and how their probabilities are arranged.

NumPy supports several random distributions that are useful for simulations, statistics and Machine Learning.

9.1 Uniform Distribution

- Uniform distribution generates values where every value within the specified interval has equal probability density.

```python
rng = np.random.default_rng(42)

uniform_values = rng.uniform(1, 10, size=5)

print(uniform_values)
```

# 9.2 Normal Distribution

Normal distribution is a probability distribution in which data values tend to cluster around the mean, forming a bell-shaped curve when the distribution is approximately normal.

Syntax

np.random.normal(loc, scale, size)

- loc – mean.

- scale – standard deviation.

- size – number of samples.

```python
rng = np.random.default_rng(42)

normal_values = rng.normal(loc=60, scale=10, size=10)

print(normal_values)
```

9.3 Binomial Distribution

Binomial distribution models the number of successes in a fixed number of independent trials, where each trial has the same probability of success.

Syntax

np.random.binomial(n, p, size)

- n – number of trials.
- p – probability of success.
- size – number of samples.

```python
rng = np.random.default_rng(42)

binomial_values = rng.binomial(n=10, p=0.5, size=5)

print(binomial_values)
```

9.4 Poisson Distribution

Poisson distribution models the number of events occurring within a fixed interval when events occur at an average rate.

Syntax

np.random.poisson(lam, size)

- lam – average event rate.
- size – number of samples.

```python
rng = np.random.default_rng(42)

poisson_values = rng.poisson(lam=5, size=10)

print(poisson_values)
```

9.5 Exponential Distribution

Exponential distribution models the time between consecutive events in a process where events occur at a constant average rate.

Syntax

np.random.exponential(scale, size)

- scale – mean waiting time.
- size – number of samples.

```python
rng = np.random.default_rng(42)

exponential_values = rng.exponential(scale=2, size=5)

print(exponential_values)
```

# 10. Visualizing Random Distributions

```python
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

uniform_values = rng.uniform(0, 100, size=1000)

plt.hist(uniform_values, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

# 11. Practical Example – Random Student Marks Generation

We can generate random marks for students, select a random group, and shuffle the marks.

```python
rng = np.random.default_rng(42)

student_marks = rng.integers(35, 101, size=10)

print("Student Marks:", student_marks)

selected_marks = rng.choice(student_marks, size=5, replace=False)

print("Selected Marks:", selected_marks)

shuffled_marks = rng.permutation(student_marks)

print("Shuffled Marks:", shuffled_marks)
```

Interpretation
- rng.integers() generates random student marks from 35 to 100.
- rng.choice() randomly selects five marks without replacement.
- rng.permutation() returns the marks in a randomized order without changing the original array.

Real-World Applications
- Real-world: Randomly select students for an experiment or survey.
- Business: Randomly select customers for feedback surveys or product testing.
- AI/ML: Randomly sample observations, shuffle training data, simulate data and generate test inputs.

# 12. Best Practices
- Use np.random.default_rng() for new NumPy code.
- Use a fixed seed when reproducibility is required.
- Use np.random.choice() for random sampling.
- Use replace=False when duplicate selections are not allowed.
- Use np.random.permutation() when you need a shuffled copy.
- Use np.random.shuffle() when you want to modify the original array.
- Understand the parameters of each distribution.
- Use random data for testing and simulation when appropriate.
- Avoid relying on random outputs without controlling the generator when reproducibility matters.

# 13. Advantages
- Generates random numbers efficiently.
- Supports multiple probability distributions.
- Useful for simulations and experiments.
- Supports random selection and sampling.
- Helps shuffle datasets.
- Useful for Machine Learning training and testing.
- Supports reproducible experiments with seeds.

# 14. Limitations
- Random values may not represent real-world data accurately.
- Random sampling can introduce sampling bias if used incorrectly.
- Random results may vary without a fixed seed.
- Generated random values are not automatically suitable for every experiment.
- Understanding the distribution parameters is necessary.
