# 2. Import NumPy

```python
import numpy as np

print("NumPy imported successfully")
```

# 3. Create a NumPy Array

```python
marks = np.array([85, 90, 78, 92, 88])

print(marks)
```

We will use this array throughout the notebook to perform mathematical operations.

# 4. Sum – np.sum()

np.sum() calculates the total sum of all elements in an array.


```python
marks = np.array([85, 90, 78, 92, 88])

total = np.sum(marks)

print("Total Marks:", total)
```

# 5. Mean – np.mean()

Mean is the average of all numerical values.

```python
marks = np.array([85, 90, 78, 92, 88])

mean = np.mean(marks)

print("Mean:", mean)
```

# 6. Median – np.median()

Median is the middle value of a sorted dataset. If the number of values is even, the median is the average of the two middle values.

```python
marks = np.array([85, 90, 78, 92, 88])

median = np.median(marks)

print("Median:", median)
```

# 7. Variance – np.var()

Variance measures how much numerical values are spread out from the mean.

```python
marks = np.array([85, 90, 78, 92, 88])

variance = np.var(marks)

print("Variance:", variance)
```

Sample Variance

When calculating sample variance, we use ddof=1.

```python
marks = np.array([85, 90, 78, 92, 88])

sample_variance = np.var(marks, ddof=1)

print("Sample Variance:", sample_variance)
```

# 8. Standard Deviation – np.std()

Standard deviation measures how much data values typically vary from the mean.

```python
marks = np.array([85, 90, 78, 92, 88])

std = np.std(marks)

print("Standard Deviation:", std)
```

Sample Standard Deviation

```python
marks = np.array([85, 90, 78, 92, 88])

sample_std = np.std(marks, ddof=1)

print("Sample Standard Deviation:", sample_std)
```

# 9. Minimum – np.min()

np.min() returns the smallest element in an array.

```python
marks = np.array([85, 90, 78, 92, 88])

minimum = np.min(marks)

print("Minimum Marks:", minimum)
```

# 10. Maximum – np.max()

np.max() returns the largest element in an array.

```python
marks = np.array([85, 90, 78, 92, 88])

maximum = np.max(marks)

print("Maximum Marks:", maximum)
```

# 11. Argmin – np.argmin()

np.argmin() returns the index position of the smallest element in an array.

```python
marks = np.array([85, 90, 78, 92, 88])

index = np.argmin(marks)

print("Index of Minimum:", index)
```

# 12. Argmax – np.argmax()

np.argmax() returns the index position of the largest element in an array.

```python
marks = np.array([85, 90, 78, 92, 88])

index = np.argmax(marks)

print("Index of Maximum:", index)
```

# 13. Percentile – np.percentile()

Percentile indicates the position below which a specified percentage of data values fall.

Formula

The percentile represents a value at a given percentage position in the data.

```python
marks = np.array([85, 90, 78, 92, 88])

p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

Interpretation
- 25th percentile: 85
- 50th percentile: 88
- 75th percentile: 90
The 50th percentile is the median.

# 14. Dot Product – np.dot()

The dot product multiplies corresponding elements of two arrays and adds the products together.

```python
marks = np.array([80, 90, 70])
weights = np.array([2, 3, 1])

result = np.dot(marks, weights)

print("Dot Product:", result)
```

# 15. Matrix Multiplication

Matrix multiplication combines rows of the first matrix with columns of the second matrix.

```python
matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])

result = np.matmul(matrix1, matrix2)

print(result)
```

# 16. Transpose – np.transpose()

Transpose changes rows into columns and columns into rows.

```python
matrix = np.array([[10, 20, 30], [40, 50, 60]])

result = np.transpose(matrix)

print(result)
```

Alternative Method

```python
matrix = np.array([[10, 20, 30], [40, 50, 60]])

print(matrix.T)
```

# 17. Basic Linear Algebra Operations

NumPy provides a linalg module for linear algebra operations.

Common operations include:

# Function	Purpose

np.linalg.det()	--> Calculates determinant

np.linalg.inv()	--> Calculates inverse

np.linalg.solve()-->Solves linear equations

np.linalg.norm()-->Calculates vector or matrix norm

# 17.1 Determinant – np.linalg.det()

The determinant is a scalar value calculated from a square matrix.


```python
matrix = np.array([[1, 2], [3, 4]])

determinant = np.linalg.det(matrix)

print("Determinant:", determinant)
```

# 17.2 Matrix Inverse – np.linalg.inv()

The inverse of a square matrix is a matrix that, when multiplied by the original matrix, gives the identity matrix (when the inverse exists).

```python
matrix = np.array([[1, 2], [3, 4]])

inverse = np.linalg.inv(matrix)

print(inverse)
```

# 17.3 Solve Linear Equations – np.linalg.solve()

This function solves a system of linear equations.

```python
A = np.array([[1, 2], [3, 4]])

B = np.array([5, 11])

solution = np.linalg.solve(A, B)

print("Solution:", solution)
```

# 17.4 Norm – np.linalg.norm()

Norm calculates the magnitude or length of a vector.

```python
vector = np.array([3, 4])

result = np.linalg.norm(vector)

print("Norm:", result)
```

# 18. Practical Example – Student Marks Analysis

Real-World, Business and AI/ML Applications
- Real-world example: Calculate student scores, averages, variation and performance levels.
- Business example: Calculate sales totals, average revenue, highest and lowest sales, and weighted performance scores.
- AI/ML example: Mathematical operations are used to calculate feature values, matrix multiplication, 
    numerical ransformations, model predictions and statistical measurements.

```python
import numpy as np

marks = np.array([85, 90, 78, 92, 88])

print("Total:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Variance:", np.var(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Minimum Index:", np.argmin(marks))
print("Maximum Index:", np.argmax(marks))
print("75th Percentile:", np.percentile(marks, 75))
```
