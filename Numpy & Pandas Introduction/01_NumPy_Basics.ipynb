# 1. What is NumPy?

NumPy stands for Numerical Python. It is a Python library used for numerical calculations, mathematical operations and working with arrays.

NumPy provides a powerful data structure called ndarray (N-dimensional array), which allows us to store and perform operations on numerical data efficiently.

# 2. Why NumPy?

NumPy is useful because:
- It performs mathematical operations efficiently.
- It supports multidimensional arrays.
- It uses less memory for many numerical workloads than ordinary Python lists.
- It supports vectorized operations.
- It provides mathematical functions.
- It is widely used in Data Analysis, Machine Learning and Data Science.

# 4. NumPy Arrays (ndarray)

A NumPy array is a collection of elements stored in an array structure. NumPy arrays can have one or multiple dimensions.

# 5. Creating Arrays

NumPy provides different methods to create arrays.

5.1 Create an Array from a List

```python
marks = np.array([85, 90, 78, 92, 88])

print(marks)
```

5.2 Create an Array of Zeros

```python
zero_array = np.zeros(5)

print(zero_array)
```

5.3 Create an Array of Ones

```python
one_array = np.ones(5)

print(one_array)
```

5.4 Create an Array using arange()

```python
numbers = np.arange(1, 10)

print(numbers)
```

np.arange(start, stop) generates numbers from the start value up to, but not including, the stop value.

5.5 Create an Array using linspace()

```python
numbers = np.linspace(0, 10, 5)

print(numbers)
```

np.linspace() generates a specified number of equally spaced values between two endpoints.

# 6. Dimensions of Arrays

The dimension of an array represents the number of axes in the array.

6.1 One-Dimensional Array (1D)

```python
arr_1d = np.array([10, 20, 30, 40])

print(arr_1d)
print("Dimensions:", arr_1d.ndim)
```

6.2 Two-Dimensional Array (2D)

```python
arr_2d = np.array([[10, 20, 30], [40, 50, 60]])

print(arr_2d)
print("Dimensions:", arr_2d.ndim)
```

6.3 Three-Dimensional Array (3D)

```python
arr_3d = np.array([[[10, 20], [30, 40]], [[50, 60], [70, 80]]])

print(arr_3d)
print("Dimensions:", arr_3d.ndim)
```

# 10. Indexing in NumPy

Indexing is used to access individual elements from an array.

```python
marks = np.array([85, 90, 78, 92, 88])

print(marks[0])
print(marks[2])
print(marks[-1])
```

- marks[0] accesses the first element.
- marks[2] accesses the third element.
- marks[-1] accesses the last element.

Indexing in a 2D Array

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])

print(arr[0, 1])
print(arr[1, 2])
```

- arr[0, 1] accesses the element in the first row and second column.
- arr[1, 2] accesses the element in the second row and third column.

# 11. Slicing in NumPy

Slicing is used to extract a portion of an array.

```python
marks = np.array([85, 90, 78, 92, 88])

print(marks[1:4])
print(marks[:3])
print(marks[2:])
```

Slicing in a 2D Array

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])

print(arr[:, 1])
```

# 12. Reshaping Arrays

Reshaping changes the structure of an array without changing the total number of elements.

```python
arr = np.arange(1, 7)

reshaped_arr = arr.reshape(2, 3)

print(reshaped_arr)
```


Reshape into 3 Rows and 2 Columns

```python
arr = np.arange(1, 7)

reshaped_arr = arr.reshape(3, 2)

print(reshaped_arr)
```

The total number of elements must remain the same. Here, 6 elements are reshaped into 3 rows and 2 columns.

# 13. Flatten an Array

flatten() converts a multidimensional array into a one-dimensional array and returns a copy.

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])

flat_arr = arr.flatten()

print(flat_arr)
```

The original 2D array becomes a 1D array with 6 elements.

# 14. Copy vs View

Copy

A copy creates a separate array. Changes to the copied array do not affect the original array.

```python
arr = np.array([10, 20, 30, 40])

copy_arr = arr.copy()

copy_arr[0] = 100

print("Original:", arr)
print("Copy:", copy_arr)
```

View

A view shares the underlying data with the original array in typical NumPy view operations. Changes to the view can affect the original.

```python
arr = np.array([10, 20, 30, 40])

view_arr = arr.view()

view_arr[0] = 100

print("Original:", arr)
print("View:", view_arr)
```

# 15. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of different shapes when their dimensions are compatible.
Example 1 – Adding a Scalar to an Array

```python
marks = np.array([50, 60, 70, 80])

result = marks + 5

print(result)
```

NumPy adds 5 to every element without requiring a loop.

Example 2 – Adding Two Arrays of Different Shapes

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])

add_arr = np.array([1, 2, 3])

result = arr + add_arr

print(result)
```

# 16. Array Operations

NumPy supports direct mathematical operations on arrays.

```python
marks = np.array([50, 60, 70, 80])

print("Addition:", marks + 5)
print("Subtraction:", marks - 5)
print("Multiplication:", marks * 2)
print("Division:", marks / 2)
```

Array-to-Array Operations

```python
arr1 = np.array([10, 20, 30])

arr2 = np.array([1, 2, 3])

print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

# 17. Universal Functions (ufuncs)

Universal functions (ufuncs) are NumPy functions that perform mathematical operations element-wise on arrays.
Common ufuncs

# Function	Purpose

np.sqrt() -->	Calculates square root

np.square() -->	Squares each element

np.exp() --> Calculates exponential

np.log() --> Calculates natural logarithm

np.abs() --> Calculates absolute value

np.round() -->	Rounds values

np.sin() --> Calculates sine


```python
arr = np.array([4, 9, 16, 25])

print("Square Root:", np.sqrt(arr))
print("Square:", np.square(arr))
print("Absolute:", np.abs(arr))
```

Example using np.round()

```python
arr = np.array([10.456, 20.789, 30.123])

print(np.round(arr, 2))
```

# 18. Practical Example – Student Marks Analysis

Real-World, Business and AI/ML Applications
- Real-world example: NumPy can help analyze student marks, attendance, scores and numerical performance data.
- Business example: NumPy can help calculate sales, revenue, expenses and average performance.
- AI/ML example: NumPy arrays are used to represent numerical data, prepare features, perform 
      mathematical operations and support ML calculations.

```python
import numpy as np

marks = np.array([85, 90, 78, 92, 88])

print("Student Marks:", marks)
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Passed Students:", marks >= 40)
print("Marks after adding 5:", marks + 5)
```

Output Interpretation
- np.sum() calculates the total marks.
- np.mean() calculates the average marks.
- np.max() returns the highest mark.
- np.min() returns the lowest mark.
- marks >= 40 returns Boolean values for the pass condition.
- marks + 5 adds 5 to every student’s mark.

# 19. Best Practices
- Use NumPy arrays for numerical calculations.
- Use vectorized operations instead of unnecessary Python loops.
- Check the shape and dimensions before performing operations.
- Choose suitable data types to reduce memory usage.
- Use .copy() when you need independent data.
- Use broadcasting when array dimensions are compatible.
- Avoid unnecessary conversions between lists and arrays.

# 20. Advantages of NumPy
- Fast numerical calculations.
- Efficient array operations.
- Supports multidimensional arrays.
- Provides many mathematical functions.
- Supports broadcasting.
- Useful for Data Analysis and Machine Learning.
- Can reduce memory usage for numerical data compared with ordinary Python lists.

# 21. Limitations of NumPy
- Arrays generally store elements of a common data type.
- Not ideal for handling complex tabular data with different column types compared with Pandas.
- Broadcasting can be confusing when array shapes are incompatible.
- NumPy is not designed primarily for handling missing values and labeled data in the way Pandas is.
