# 1. Introduction to Measures of Central Tendency

Measures of central tendency are statistical methods used to identify the central or typical value of a dataset.

They help us understand a large collection of data using one representative value.

The three main measures are:
1. Mean
2. Median
3. Mode

Other important measures include:
4. Weighted Mean
5. Geometric Mean
6. Harmonic Mean

Real-world example: A teacher wants to understand the typical marks scored by students. The teacher can calculate the mean, median, or mode of the students' marks.

Business example: A company calculates its average monthly sales to understand its business performance.

AI/ML example: Data scientists use central tendency measures to understand datasets and fill missing numerical values during data preprocessing.

# 2. Mean (Average)

Mean is the sum of all values divided by the total number of values.

```python
import numpy as np

marks = [10, 20, 30, 40, 50]

mean_value = np.mean(marks)

print("Mean:", mean_value)
```

AI/ML Example

In machine learning, mean is used to fill missing numerical values.

For example, if some students' marks are missing, we can calculate the average marks and use it to replace missing values.

```python
import pandas as pd
from sklearn.impute import SimpleImputer

data = pd.DataFrame({
    "Marks": [50, 60, np.nan, 80, 90]
})

imputer = SimpleImputer(strategy="mean")

data["Marks"] = imputer.fit_transform(data[["Marks"]])

print(data)
```

Interpretation

The average marks are 70. The missing value is replaced with 70.

# 3. Median

Median is the middle value of a dataset when the values are arranged in ascending or descending order.

If the dataset contains an even number of values, the median is the average of the two middle values.

```python
import numpy as np

numbers = [10, 20, 30, 40, 50]

median_value = np.median(numbers)

print("Median:", median_value)
```

AI/ML Example

Median is used to handle missing numerical values, especially when data contains outliers.

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "Salary": [20000, 25000, np.nan, 35000, 200000]
})

imputer = SimpleImputer(strategy="median")

data["Salary"] = imputer.fit_transform(data[["Salary"]])

print(data)
```

Interpretation

The missing salary is replaced with ₹30,000, the median salary.

# 4. Mode

Mode is the value that appears most frequently in a dataset.

A dataset can have:
- One mode – Unimodal
- Two modes – Bimodal
- More than two modes – Multimodal
- No repeated value – No mode

Formula

Mode is the value with the highest frequency.

Python Implementation

```python
from scipy import stats

numbers = [10, 20, 20, 30, 40]

mode_value = stats.mode(numbers, keepdims=False).mode

print("Mode:", mode_value)
```

AI/ML Example

Mode is commonly used to fill missing categorical values.

For example, if the most common product category is Electronics, missing category values can be filled with Electronics.

```python
import pandas as pd
from sklearn.impute import SimpleImputer

data = pd.DataFrame({
    "Category": ["Electronics", "Clothing", None,
                 "Electronics", "Food"]
})

imputer = SimpleImputer(strategy="most_frequent")

data["Category"] = imputer.fit_transform(data[["Category"]]).ravel()

print(data)
```

Interpretation

The missing category is replaced with Electronics because it is the most frequently occurring category.

# 5. Weighted Mean

Weighted Mean is an average in which different values are given different levels of importance or weights.

It is useful when all values do not have equal importance.

```python
import numpy as np

marks = [80, 70, 90]
weights = [3, 2, 1]

weighted_mean = np.average(marks, weights=weights)

print("Weighted Mean:", weighted_mean)
```

AI/ML Example

Geometric mean can be used to combine performance measures when a multiplicative average is appropriate, such as comparing multiple model performance ratios.

Interpretation

The geometric mean is useful when values represent growth factors or proportional changes rather than simple additive quantities.

# 7. Harmonic Mean

Harmonic Mean is the reciprocal of the arithmetic mean of the reciprocals of the values.

It is especially useful for averaging rates and ratios.

```python
import numpy as np

speeds = [40, 60]

harmonic_mean = len(speeds) / np.sum(1 / np.array(speeds))

print("Harmonic Mean:", harmonic_mean)
```

AI/ML Example

The F1-score is the harmonic mean of precision and recall.


```python
from sklearn.metrics import f1_score

actual = [1, 1, 0, 0, 1, 0]
predicted = [1, 0, 0, 0, 1, 1]

score = f1_score(actual, predicted)

print("F1 Score:", score)
```

# 8. Important Differences
- Mean is affected by extreme values.
- Median is less affected by extreme values.
- Mode identifies the most frequently occurring value.
- Weighted mean considers different importance levels.
- Geometric mean is appropriate for positive multiplicative data.
- Harmonic mean is appropriate for certain rates and ratios.

# 9. Python Practical Implementation

Now calculate all six measures using one dataset.

```python
import numpy as np
from scipy import stats

# Dataset
numbers = [10, 20, 20, 30, 40]

# Mean
mean_value = np.mean(numbers)

# Median
median_value = np.median(numbers)

# Mode
mode_value = stats.mode(numbers, keepdims=False).mode

# Weighted Mean
values = [80, 70, 90]
weights = [3, 2, 1]

weighted_mean = np.average(values, weights=weights)

# Geometric Mean
geometric_mean = stats.gmean([2, 4, 8])

# Harmonic Mean
harmonic_mean = stats.hmean([40, 60])

# Display results
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
print("Weighted Mean:", weighted_mean)
print("Geometric Mean:", geometric_mean)
print("Harmonic Mean:", harmonic_mean)
```

Interpretation of Results
- Mean is 24, which represents the average value.
- Median is 20, which represents the middle value.
- Mode is 20, which occurs most frequently.
- Weighted mean is 78.33, considering different weights.
- Geometric mean is 4, representing the multiplicative average.
- Harmonic mean is 48, representing the average of the two rates in this example.

# 10. Data Visualization

We can visualize the mean, median, and mode using a histogram.

```python
import matplotlib.pyplot as plt
import numpy as np

numbers = [10, 20, 20, 30, 40]

mean_value = np.mean(numbers)
median_value = np.median(numbers)

plt.hist(numbers, bins=5, edgecolor="black")

plt.axvline(mean_value, linestyle="--",
            label="Mean")

plt.axvline(median_value, linestyle="-.",
            label="Median")

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Measures of Central Tendency")

plt.legend()
plt.show()
```

Expected Result:

A histogram will display the frequency distribution of the values, with vertical lines representing the mean and median.

Interpretation

The histogram helps us understand the distribution of data and compare the mean and median visually.

# 11. Advantages of Measures of Central Tendency

1. Summarize large datasets using a single value.
2. Make data easier to understand.
3. Help compare different datasets.
4. Support business decision-making.
5. Help in data analysis and preprocessing.
6. Useful in machine learning and statistics.
12. Limitations of Measures of Central Tendency
1. Mean can be affected by outliers.
2. Median does not use the exact magnitude of every value.
3. Mode may not exist or may not be unique.
4. A single central value cannot describe the entire dataset.
5. Different datasets can have the same mean but different distributions.
6. Geometric and harmonic means are not suitable for every type of data.
