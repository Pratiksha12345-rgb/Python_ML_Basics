# 1. Introduction to Hypothesis Testing

What is Hypothesis Testing?

Hypothesis Testing is a statistical method used to make decisions or draw conclusions about a population using sample data.

It helps us test whether an assumption or claim about data is supported by evidence.

Example: Suppose a college claims that the average student mark is 60. We collect marks from a sample of students and use Hypothesis Testing to check whether the average mark is actually 60.

Why is Hypothesis Testing Important?
- Helps make decisions using data.
- Helps test assumptions.
- Helps compare groups.
- Helps identify statistically significant differences.
- Helps avoid decisions based only on assumptions.
- Used in data analysis and AI/ML research.

# 2. What is a Hypothesis?

A Hypothesis is a statement or assumption about a population that we want to test using statistical data.

In Hypothesis Testing, we usually define two hypotheses.

A. Null Hypothesis (H₀)

The Null Hypothesis represents the default assumption. It usually states that there is no significant difference, effect or change.

Example: The average student mark is 60.
\[
H_0:\mu=60
\]

B. Alternative Hypothesis (H₁)

The Alternative Hypothesis represents the claim that we want to investigate. It states that a difference, effect or change exists.

Example: The average student mark is not 60.
\[
H_1:\mu\neq60
\]

# Difference Between Null and Alternative Hypothesis

Null Hypothesis (H₀)	Alternative Hypothesis (H₁)

Default assumption.	Opposes the null hypothesis.

Usually states no difference or effect.	States a difference or effect exists.

Tested using statistical evidence.	Supported when evidence against H₀ is sufficient.


# 3. Steps of Hypothesis Testing

Hypothesis Testing generally follows these steps:
1. Define the null hypothesis.
2. Define the alternative hypothesis.
3. Choose the significance level.
4. Select an appropriate statistical test.
5. Calculate the test statistic and p-value.
6. Compare the p-value with the significance level.
7. Draw a conclusion.

Example

Suppose a college claims that the average marks of students are 60.

We collect a sample of student marks and test the claim.
- H₀: Average marks are 60.
- H₁: Average marks are not 60.
- Significance level: 0.05.

We will perform the test using Python later in this notebook.

# 4. Significance Level (Alpha)

What is Significance Level?

The Significance Level is the threshold used to determine whether the statistical evidence is strong enough to reject the Null Hypothesis.

```python
alpha = 0.05

print("Significance Level:", alpha)
```

# 5. What is a P-Value?

The P-value is the probability, assuming the Null Hypothesis is true, of obtaining a test result at least as extreme as the one observed.

It helps us determine whether the observed result provides evidence against the Null Hypothesis.

Decision Rule

Condition	Decision

P-value ≤ 0.05	Reject the Null Hypothesis.
P-value > 0.05	Fail to reject the Null Hypothesis.

Important: We do not say that we prove or accept the Null Hypothesis. We either reject it or fail to reject it based on the evidence.

Example
- P-value = 0.02
- Significance level = 0.05

Since 0.02 is less than 0.05, we reject the Null Hypothesis.

If:
- P-value = 0.08
- Significance level = 0.05

Since 0.08 is greater than 0.05, we fail to reject the Null Hypothesis.

# 6. Type I and Type II Errors

What is Type I Error?

Type I Error occurs when we reject a true Null Hypothesis.

It is also called a False Positive.

Example: We conclude that the average marks are different from 60 when the true population average is actually 60.

What is Type II Error?

Type II Error occurs when we fail to reject a false Null Hypothesis.

It is also called a False Negative.

Example: We conclude that there is insufficient evidence that average marks differ from 60 when the true population average is actually different from 60.

Comparison

Type	Meaning	Error
Type I Error	Rejecting a true H₀.	False Positive.
Type II Error	Failing to reject a false H₀.	False Negative.


# 7. One-Tailed and Two-Tailed Tests

One-Tailed Test

A One-Tailed Test checks whether a parameter is significantly greater than or less than a specified value in one direction.

Example: We want to test whether the average student mark is greater than 60.
\[
H_0:\mu=60
\]

\[

H_1:\mu>60
\]

Types of One-Tailed Tests
- Right-Tailed Test: Checks whether a value is greater than a specified value.
- Left-Tailed Test: Checks whether a value is less than a specified value.

Two-Tailed Test
A Two-Tailed Test checks whether a parameter is significantly different from a specified value in either direction.

Example: We want to test whether the average student mark is different from 60.
\[
H_0:\mu=60
\]

\[
H_1:\mu\neq60
\]

Difference

One-Tailed Test	    Two-Tailed Test

Tests one direction.	Tests both directions.

Greater than or less than.	Different from.
