# Notebook 1 – Introduction to Machine Learning

# 1. What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence that allows computers to learn from data and make predictions or decisions without being explicitly programmed for every task.

Example: An email system learns from previous emails to identify whether a new email is spam or not.

# 2. Why Machine Learning?

Machine Learning is used because:

It can learn patterns from large amounts of data.
It can make predictions based on previous information.
It can automate repetitive tasks.
It can improve performance with more relevant training data.
It helps businesses make data-driven decisions.

# Real-world applications:

Fraud detection in banking.
Product recommendations in shopping applications.
Disease prediction in healthcare.
Spam email detection.
Customer behaviour prediction.

# 4. Traditional Programming vs Machine Learning

Traditional Programming: Programmers provide rules and data to produce an output.

Machine Learning: The system learns patterns from training data and uses them to make predictions.

# Example:

Traditional programming: Give fixed rules to identify spam emails.

Machine Learning: Train a model using examples of spam and non-spam emails.

# 5. Types of Machine Learning

# 5.1 Supervised Learning

Supervised Learning uses labelled data. The model learns from input features and their correct target values.

Examples:

House price prediction.

Spam email classification.

Student result prediction.

# 5.2 Unsupervised Learning

Unsupervised Learning uses data without labelled target values. The model finds hidden patterns or groups.

Examples:

Customer segmentation.

Grouping similar products.

Finding unusual patterns.

# 5.3 Semi-Supervised Learning

Semi-Supervised Learning uses a small amount of labelled data and a large amount of unlabelled data.

Example: Image classification when only a few images have labels.

# 5.4 Reinforcement Learning

Reinforcement Learning involves an agent learning by interacting with an environment and receiving rewards or penalties.

Examples:

Game-playing systems.

Robot navigation.

Self-learning control systems.

# 6. Classification, Regression and Clustering

Classification: Predicts a category or class.

Example: Predict whether an email is spam or not spam.

Regression: Predicts a continuous numerical value.

Example: Predicting house prices.

Clustering: Groups similar data points without predefined labels.

Example: Grouping customers based on purchasing behaviour.

# 7. Important Machine Learning Terms

Features: Input variables used by a model to make predictions.

Target: The output variable that the model tries to predict.

Training: The process of teaching a model using training data.

Prediction: The output generated by a trained model.

Model: A learned mathematical or computational representation of patterns in data.

Parameters: Values learned by the model during training.

Hyperparameters: Settings selected before or during model training to control the learning process.

# 8. Complete Machine Learning Workflow

Define the problem.

Collect the data.

Understand the data.

Clean and preprocess the data.

Perform feature engineering.

Split the dataset into training and testing sets.

Select an appropriate algorithm.

Train the model.

Make predictions.

Evaluate model performance.

Improve the model.

Deploy the model.

# 9. Simple Python Example

We will use Linear Regression to predict a student's marks based on study hours.

import numpy as np
from sklearn.linear_model import LinearRegression

# Input feature: study hours
X = np.array([[1], [2], [3], [4], [5]])

# Target: student marks
y = np.array([20, 30, 40, 50, 60])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict marks for 6 study hours
prediction = model.predict([[6]])

print("Predicted Marks:", prediction[0])

# Expected Output:

Predicted Marks: 70.0

# Explanation:

X contains the study hours.

y contains the actual marks.

LinearRegression() creates the model.

fit() trains the model using the given data.

predict() predicts marks for 6 study hours.

The model learns a linear relationship between study hours and marks.
